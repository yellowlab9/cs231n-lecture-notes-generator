However, defining the pretext task itself is not straightforward; there are many different ways of defining those. We want to define the pretext task in a way that it is first general enough that it can get us good features and doesn't require manual labeling. The labels should come from the data itself. For example, one option would be image completion, where we give parts of an image that are unmasked and define a task to predict the masked parts.

Another example is rotating the image with a specific angle; the task is to take the image as input and predict what was the rotation angle it went through. We could also use a jigsaw puzzle, where we have patches of the image that are out of order, but the task is to output the correct order of these patches.
